```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1. Load data
df = pd.read_csv("/content/shopify_stock.csv")
df
```

```python
plt.plot(df['date'], df['close'])

plt.xlabel('Date')
plt.ylabel('Closing Price')
plt.title("Shopify's Closing Price Over Time")

plt.show()
```

```python
# Calculate moving averages
df['MA20'] = df['close'].rolling(20).mean()
df['MA50'] = df['close'].rolling(50).mean()

# Plot
plt.plot(df['close'], label='Closing Price')
plt.plot(df['MA20'], label='20 Days MA')
plt.plot(df['MA50'], label='50 Days MA')

plt.xlabel('Days')
plt.ylabel('Price')
plt.title("Shopify Closing Price with Moving Averages")
plt.legend()
plt.show()
```

```python
import seaborn as sns
# Calculate daily return
df['Daily_Return'] = df['close'].pct_change()

# Histogram + KDE plot
sns.histplot(df['Daily_Return'].dropna(), kde=True)

plt.xlabel('Daily Return')
plt.ylabel('Frequency')
plt.title('Daily Return Distribution')
plt.show()
```
